```python
# while loop
count=0
while count<5:
    print("Hello")
    count+=1
```

```python
# break keyword
c=0
while True:
    print("Hello")
    c+=1
    if c==2:
        print("bye")
        break
```

```python
# continue keyword
c=0
while c<5:
    if c==2:
        c+=1
        continue
    print(c,end=" ")
    c+=1
```

```python
# Reverse a number
a=int(input("Enter number: "))
a=str(a)
res=""
for i in range(len(a)-1,-1,-1):
    res+=a[i]
a=int(res)
print(a)
```

```python
# Find the sum of digits in a number
a=int(input("Enter number: "))
a=str(a)
res=0
for i in a:
    res+=int(i)
print(res)
```

```python
# Righht angled triangle
a=int(input("Enter height: "))
for i in range(a):
    for j in range(i+1):
        print("*",end="")
    print()
```

```python
# check palindrome number
a=int(input("Enter number: "))
a=str(a)
i,j=0,len(a)-1
res=True
while i<=j:
    if a[i]!=a[j]:
        res=False
        break
    i+=1
    j-=1
print(res)
```
