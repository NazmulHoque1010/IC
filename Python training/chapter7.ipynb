```python
# initialize dictionary
my_cat = {'size': 'fat', 'color': 'gray', 'age': 17}
print(my_cat)
```

```python
# value updating
my_cat["color"]="white"
print(my_cat)
```

```python
# order
spam = ['cats', 'dogs', 'moose']
bacon = ['dogs', 'moose', 'cats']
print(spam == bacon)  # The order of list items matters.

eggs = {'name': 'Zophie', 'species': 'cat', 'age': '8'}
ham = {'species': 'cat', 'age': '8', 'name': 'Zophie'}
print(eggs == ham)  # The order of dictionary key-value pairs doesn't matter.
```

```python
# keys and values
spam = {'color': 'red', 'age': 42}
for v in spam.values():
    print(v)

for k in spam.keys():
    print(k)

print(spam.keys())
print(spam.values())
print(list(spam.keys()))  # Returns proper list form
```

```python
# items()
for k, v in spam.items():
    print('Key: ' + str(k) + ' Value: ' + str(v))

print(list(spam.items()))
```

```python
# checking availability of a key in a dictionary using the .get() method
picnic_items = {'apples': 5, 'cups': 2}
print('I am bringing ' + str(picnic_items.get('cups', 0)) + ' cups.')
print('I am bringing ' + str(picnic_items.get('eggs', 0)) + ' eggs.')
```

```python
>>> 'I am bringing ' + str(picnic_items['eggs']) + ' eggs.'
```

```python
spam = {'name': 'Pooka', 'age': 5}
spam.setdefault('color', 'black')  # Sets 'color' key to 'black'
print(spam)
spam.setdefault("color","white") # does nothing
print(spam["color"])
```

```python
message = 'It was a bright cold day in April, and the clocks were striking thirteen.'
count={}
for i in message:
    count.setdefault(i,0)
    count[i]+=1
print(count)
```
