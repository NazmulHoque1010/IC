```python
# making a series of files and folders
base_dir=Path(r'C:\Users\Lenovo\Desktop\IC\Python training\ch11')
(base_dir / 'spam').mkdir(exist_ok=True)
(base_dir / 'spam/eggs').mkdir(exist_ok=True)
(base_dir / 'spam/eggs2').mkdir(exist_ok=True)
(base_dir / 'spam/eggs/bacon').mkdir(exist_ok=True)
for f in ['spam/file1.txt', 'spam/eggs/file2.txt', 'spam/eggs/file3.txt','spam/eggs/bacon/file4.txt']:
    with open(base_dir / f, 'w', encoding='utf-8') as file:
        file.write('Hello')
```

```python
h = base_dir

for folder_name, subfolders, filenames in os.walk(h / 'spam'):
    print('The current folder is ' + folder_name)

    for subfolder in subfolders:
        print('SUBFOLDER OF ' + folder_name + ': ' + subfolder)

    for filename in filenames:
        print('FILE INSIDE ' + folder_name + ': '+ filename)
        # Rename file to uppercase:
        p = Path(folder_name)
        shutil.move(p / filename, p / filename.upper())
   
    print('')
```

```python
import zipfile
with open(base_dir/'file1.txt','w',encoding='utf-8') as file_obj:
    file_obj.write("Hello"*10000)
with zipfile.ZipFile(base_dir/'example_zip','w') as example_zip:
    example_zip.write(base_dir/'file1.txt',compress_type=zipfile.ZIP_DEFLATED,compresslevel=9)
```

```python
ex=zipfile.ZipFile(base_dir/'example_zip')
print(ex.namelist())
file1_info=ex.getinfo(base_dir/'file1.txt')
file1_info.file_size
```
