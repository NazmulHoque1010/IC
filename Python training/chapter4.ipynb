```python
def spam():
    eggs = 'spam local'
    print(eggs)  # Prints 'spam local'

def bacon():
    eggs = 'bacon local'
    print(eggs)  # Prints 'bacon local'
    spam()
    print(eggs)  # Prints 'bacon local'

eggs = 'global'
bacon()
print(eggs)  # Prints 'global'
```

```python
# Global variables
def spam():
    global eggs
    eggs = 'spam'

eggs = 'global'
spam()
print(eggs)  # Prints 'spam'
```

```python
# Error handling
def divide_by(num):
    try:
        return 42/num
    except ZeroDivisionError:
        print("Invalid input")
print(divide_by(3))
print(divide_by(0))
```
